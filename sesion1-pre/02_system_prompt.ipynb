# Nivel 2 — System prompt
Mismo mensaje que en el nivel 1, con y sin un rol definido en el `system`.

```python
import os

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()  # lee OPENROUTER_API_KEY de practicas/.env

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],
)

MODELO = "openai/gpt-4o-mini"
MENSAJE = "Hola, preséntate en una frase."

def preguntar(mensaje, system=None):
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": mensaje})
    r = client.chat.completions.create(model=MODELO, messages=messages)
    return r.choices[0].message.content
```

## Sin system prompt (nivel 1)

```python
print(preguntar(MENSAJE))
```

## Con rol: experto en estimación de software

```python
SYSTEM = """Eres un experto en estimación de proyectos de software con 15 años de experiencia.
Respondes de forma profesional y estructurada, y siempre relacionas la conversación
con la planificación, los riesgos y la estimación de esfuerzo."""

print(preguntar(MENSAJE, system=SYSTEM))
```

## Extra: pregunta del dominio

```python
PREGUNTA = "¿Cuánto se tarda en hacer una app de login?"
print("--- SIN SYSTEM ---\n", preguntar(PREGUNTA))
print("\n--- CON SYSTEM ---\n", preguntar(PREGUNTA, system=SYSTEM))
```
